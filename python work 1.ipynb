```python
#Take a 3-digit number and print sum of its digits.
num= int(input("enter the 3-digit number: "))

hundreds= num//100         # Gets the first digit
tens= (num//10)%10          # Gets the middle digit
ones= num%10                  # Gets the last digit

digits_sum= hundreds+tens+ones
print(f"The sum of 3 digits is: {digits_sum}")
```

```python
#Remove all spaces from a string.
string= "p y t h o n"
string= string.replace(" ", "")
print(string)

#another way
string= "p y t h o n"
string= string.split()  # Split the string into a list of words
string= "".join(string)  # Join the list back into a string without spaces
print(string)

```

```python
#Count number of vowels and consonants.
text= "hello world"
vowel= "aeiouAEIOU"
v_count= 0
c_count= 0
for char in text:
    if char.isalpha():    #donot count space
        if char in vowel:
            v_count +=1
        else:
            c_count+=1
print(f"Vowels: {v_count}, Consonants: {c_count}")


```

```python
#Find Length of String Without Using len().
text= "python"
count=0
for char in text:
    count+=1

print(f"length: {count}")
```

```python
#Replace All Vowels with *.
letter= "Beautiful"
vowels= "aeiouAEIOU"
result= "" 

for char in letter:
    if char in vowels:
        result += "*"
    else:
        result+= char
print(result)
```
